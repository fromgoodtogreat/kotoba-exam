```python
import pandas as pd

# 1. ダウンロードしたCSVを読み込む
df = pd.read_csv('words.csv')

# 2. アプリが認識しやすいように列名を変更する
# （※元のスプレッドシートが左から「語彙」「意味」「例文」の順に並んでいる前提です）
df.columns = ['word', 'meaning', 'example']

# 3. 欠損値（空欄）を空文字列に変換
df = df.fillna('')

# 4. JSONファイルとして書き出す
df.to_json('words.json', orient='records', force_ascii=False, indent=2)
print("words.json の作成が完了しました！")
```
