```python
!pip install pykakasi

import json
import re
import pykakasi

kks = pykakasi.kakasi()

def to_ruby_html(text):
    if not text:
        return ""
    # 既存の丸括弧の読み（例: 怪訝（けげん））を先に <ruby> に変換
    text = re.sub(
        r'([一-龠々〆ヵヶ]+)[（\(]([ぁ-んァ-ヶー]+)[）\)]',
        r'<ruby>\1<rt>\2</rt></ruby>',
        text
    )
    # すでに <ruby> タグがある部分とそれ以外を分割して処理
    parts = re.split(r'(<ruby>.*?</ruby>)', text)
    out = []
    for p in parts:
        if p.startswith('<ruby>'):
            out.append(p)
        else:
            for item in kks.convert(p):
                orig = item['orig']
                hira = item['hira']
                # 漢字を含むトークンのみ <ruby> を付与（送り仮名を綺麗に分離）
                if re.search(r'[一-龠々〆ヵヶ]', orig) and orig != hira:
                    out.append(f"<ruby>{orig}<rt>{hira}</rt></ruby>")
                else:
                    out.append(orig)
    return "".join(out)

with open('words.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

for row in data:
    w = re.sub(r'[（\(].*?[）\)]', '', row.get('word', '')).strip()
    row['word_ruby'] = to_ruby_html(row.get('word', ''))
    row['meaning_ruby'] = to_ruby_html(row.get('meaning', ''))
    row['example_ruby'] = to_ruby_html(row.get('example', ''))

with open('words.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("全データのふりがな付き words.json の作成が完了しました！")
```
